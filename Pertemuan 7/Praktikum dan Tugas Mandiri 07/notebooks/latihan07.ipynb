```python
import pandas as pd

# Membaca file csv menggunakan pandas
df = pd.read_csv("/content/drive/MyDrive/Hisyam Wildan Alfath/Semester 7/Machine Learning/Pertemuan 7/Praktikum dan Tugas Mandiri 07/data/dataset_satelit.csv")

# Convert 'Mg' column to numeric, coercing errors
df['Mg'] = pd.to_numeric(df['Mg'], errors='coerce')

# Fill any resulting NaN values in 'Mg' with the mean
df['Mg'].fillna(df['Mg'].mean(), inplace=True)

# Cetak header data (5 baris data) dari file
display(df.head())
```

```python
# Check for missing values before filling
print("Missing values before filling:")
display(df.isnull().sum())
```

```python
# Fill missing value in 'K' column with the mean
df['K'].fillna(df['K'].mean(), inplace=True)
```

```python
# Check for missing values after filling
print("Missing values after filling:")
display(df.isnull().sum())
```

```python
from sklearn.model_selection import train_test_split

# Use the cleaned DataFrame for splitting
X = df[['P', 'K', 'Ca', 'Mg', 'Fe', 'Mn',
       'Cu', 'Zn', 'B', 'b12', 'b11', 'b9', 'b8a', 'b8', 'b7', 'b6', 'b5',
       'b4', 'b3', 'b2', 'b1', 'Sigma_VV', 'Sigma_VH', 'plia', 'lia', 'iafe',
       'gamma0_vv', 'gamma0_vh', 'beta0_vv', 'beta0_vh']]
y = df["N"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=7
)
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X_train, y_train)
```

```python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

# Asumsikan kamu sudah punya model:
# model = LinearRegression().fit(X_train, y_train)
# dan sudah ada X_test, y_test

# Prediksi nilai N
y_pred = model.predict(X_test)

# Evaluasi model regresi
r2 = r2_score(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)

# Cetak hasil evaluasi
print("Koefisien model (pengaruh tiap fitur):")
for name, coef in zip(X_train.columns, model.coef_):
    print(f"  {name:>12}: {coef:.6f}")

print("\nIntersep (bias):      ", round(model.intercept_, 6))
print("R² (test):            ", round(r2, 4))
print("MAE:                  ", round(mae, 4))
print("RMSE:                 ", round(rmse, 4))
```

```python
slope = model.coef_[0]
intercept = model.intercept_

print(f"Persamaan: y = {slope:.2f} * x {intercept:.2f}")
```

```python
# Scatter plot antara nilai aktual dan prediksi
import matplotlib.pyplot as plt

plt.figure(figsize=(6,6))
plt.scatter(y_test, y_pred, color='purple', alpha=0.7)
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--', lw=2)

plt.xlabel("Nilai Aktual Nitrogen (N)")
plt.ylabel("Nilai Prediksi Nitrogen (N)")
plt.title("Perbandingan Prediksi vs Nilai Aktual (Model Regresi)")

plt.text(
    0.05, 0.9,
    f"$R^2$ = {r2:.3f}\nRMSE = {rmse:.3f}\nMAE = {mae:.3f}",
    transform=plt.gca().transAxes, va="top"
)
plt.tight_layout()
plt.show()
```
